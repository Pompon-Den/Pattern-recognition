```python
import numpy as np
import matplotlib.pyplot as plt


a = int(input())
b = int(input())
np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = a + b * X + np.random.randn(100, 1)

# Функция для вычисления предсказанных значений y
def predict(X, theta):
    return X.dot(theta)

# Функция для вычисления средней квадратичной ошибки (MSE) между предсказанными и истинными Y
def mse_loss(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)

# Вычисление градиента MSE 
def gradient_mse(X, y_true, y_pred):
    return -2 / len(y_true) * X.T.dot(y_true - y_pred)

# Инициализация параметров
theta = np.random.randn(2, 1)

# Градиентный спуск
n_iterations = 400
learning_rate = 0.01 # шаг обучения  
losses = []

for i in range(n_iterations):
    #   y  
    y_pred = predict(np.c_[np.ones(len(X)), X], theta)
    
    #  Вычисление потерь
    loss = mse_loss(y, y_pred)
    losses.append(loss)
    
    # Вычисление градиента
    grad = gradient_mse(np.c_[np.ones(len(X)), X], y, y_pred)
    
    # Обновление параметров
    theta -= learning_rate * grad

print("Оптимальные параметры:", theta)
print("Минимальная ошибка:", np.min(losses))
```

```python
plt.figure(figsize=(10, 6))
plt.plot(range(n_iterations), losses, label="Средняя квадратичная ошибка")
plt.xlabel('Количество итераций')
plt.ylabel('Ошибка (MSE)')
plt.title('Изменение ошибки на протяжении итераций')
plt.legend()
plt.show()
```
